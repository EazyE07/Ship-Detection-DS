```python
import cv2
from ultralytics import YOLO
from matplotlib import pyplot as plt

# 1. Das Standard-Modell laden (lädt beim ersten Mal ca. 6MB runter)
model = YOLO('yolov8n.pt') 

# 2. Ein Testbild von einem Schiff (Internet-URL)
bild_url = "https://www.inside-digital.de/img/schluss-mit-teurem-strom-und-netzproblemen-ein-schiff-soll-alles-loesen.jpg?class=1200x900"

# 3. KI drüberlaufen lassen
results = model.predict(source=bild_url, conf=0.25)

# 4. Ergebnis im Notebook anzeigen
for r in results:
    bild_mit_boxen = r.plot()
    plt.figure(figsize=(10, 6))
    plt.imshow(cv2.cvtColor(bild_mit_boxen, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
    
```
