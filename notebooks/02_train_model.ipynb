```python
from ultralytics import YOLO
import os

# 1. Modell laden
# Wir nehmen das 'nano' Modell (yolo11n.pt oder yolov8n.pt). 
# Es ist extrem schnell und reicht für die Unterscheidung von Schiffstypen völlig aus.
model = YOLO('yolo11n.pt') 

# 2. Training starten
# Wir nutzen moderate Parameter, damit dein Laptop nicht überhitzt
results = model.train(
    data='../data/processed/ship_data/data.yaml', # Pfad zur YAML Datei
    epochs=50,         # 50 Durchläufe sind ein guter Start für eine 1.0
    imgsz=640,         # Bildgröße aus dem Datensatz
    batch=16,          # Wie viele Bilder gleichzeitig verarbeitet werden (8 oder 16 ist gut für Laptops)
    name='ship_monitor_v1', # Name des Ordners für die Ergebnisse
    device='cpu'       # Wenn du keine NVIDIA Grafikkarte hast, nutze 'cpu'
)
```
